### 모듈 불러오기  
  
② from module_name import class or methods  
• from socket import gethostname  
• import된 class 또는 methods 사용할 수 있다    

```python
>>> from socket import gethostname
>>> gethostname() class or methods 이름을 직접 호출
'YKSUH'
```

③ from module_name import *  
• from socket import *  
• socket 모듈의 모든 속성을 사용할 수 있다  

```python
>>> from socket import *
>>> gethostname() 메소드를 직접 호출
'YKSUH'
```

• 편리하지만 메모리를 많이 차지하고 다른 함수와 혼동 가능성이 있다!

```python
>>> import socket as so 모듈을 단축이름으로 불러오기
>>> so.gethostname() 단축이름으로 속성 호출
'YKSUH'
```

• 다음 파일을 myModule.py로 저장  

```python
def sum(n):
 sum = 0
 for i in range(1, n+1):
    sum = sum + i
 return sum
def power(x, n):
 prod = 1
 for i in range(1, n+1):
    prod = prod * x
 return prod
if __name__ == '__main__':   # myModule.py를 직접 호출할 때만 실행됨
 print(sum(5))
 print(power(2, 3))
```

❖ if __name__ == '__main__':의 기능

```python
if __name__ == '_main__': # 모듈로 사용될 때는 실행되지 않음
    print(sum(5))
    print(power(2, 3))
```

❖ __init__.py의 역할  
• 파이썬 패키지로 인식:  
- __init__.py 파일은 해당 디렉토리를 Python 패키지로 인식하게 만듭니다.  
- Python 3.3 이상에서는 __init__.py가 없어도 패키지로 인식되지만, 명시적으로 포함하는 것이 권장됩니다.  
  
• 초기화 코드 실행:  
- 패키지가 import될 때 실행할 초기화 코드를 포함할 수 있습니다.  
- 예를 들어, 패키지 내에서 공통적으로 사용할 설정이나 변수 초기화를 수행할 수 있습니다.  
  
• __init__.py의 위치  
- mypackage/  
- │  
- ├── __init__.py # 패키지 초기화 파일  
- ├── module1.py  
- └── module2.py  

❖ 모듈 (Module)  
• 정의: Python에서 하나의 .py 파일이 하나의 모듈

❖ 패키지 (Package)  
• 정의: Python에서 디렉토리가 패키지  
- 패키지는 여러 모듈을 포함할 수 있음  
- 대규모 프로젝트에 필요  
- 디렉토리에 __init__.py 가 필수적 (비어 있더라도 인식)

파이썬 가상환경 (virtual environment)  
  
❖ 가상환경이란?  
• - 프로젝트별로 독립된 파이썬 환경을 만드는 도구  
• - 각 프로젝트가 서로 다른 패키지 버전을 사용할 수 있게 함  
• - 시스템 파이썬 환경을 오염시키지 않음  
  
❖ 만드는 방법  
• 1. venv (파이썬 3.3+ 내장 모듈)  
• 2. conda (anaconda)  
• 3. virtualenv (써드 파티 도구)  

❖ Venv 가상환경 ➔ 간단히 사용하고 싶을 경우에 사용  
• 파이선 3.3 이후부터 내장됨  

```python
# 가상환경 생성 (.venv 폴더 생성되고 그 아래 모든 정보가 저장됨)
python –m venv .myvenv (python –m venv --help,--copies, --clear, --upgrade, --prompt)
# Windows에서 활성화
.\.venv\Scripts\activate # ➔ (.venv) 프롬프트 생김
# 활성화 시 오류가 발생하면 현재 실행 정책 확인
Get-ExecutionPolicy
# 현재 사용자에 대해서만 실행 정책 변경 (최초 1번)
Set-ExecutionPolicy -ExecutionPolicy RemoteSigned -Scope CurrentUser
# macOS/Linux에서 활성화
# source myenv/bin/activate
# 비활성화 (모든 OS 공통)
deactivate
```

가상환경 및 모듈 자동 설치

❖ 파이썬 가상환경   
• 모듈, 모듈의 버전 등을 보존하여 버전 변경으로 인한 소스코드의 오류를 방지  
• 가상환경을 만들고 설치한 이후에 해당 모듈은 해당 가상환경에 저장됨  
• 설치된 모듈의 목록은 pip freeze > requirements.txt 로 저장  
• 다른 가상환경에서 해당 목록을 자동 설치할 때에는 pip install -r requirements.txt 로 설치  


```python
# 1. 현재 설치된 패키지들을 requirements.txt로 저장
pip freeze > requirements.txt
# 2. requirements.txt에서 패키지들을 설치 (올바른 명령어)
pip install -r requirements.txt
```
